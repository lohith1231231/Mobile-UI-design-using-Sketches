```python
import cv2
import numpy as np
import torch
import easyocr
import json
from segment_anything import SamPredictor, sam_model_registry

# Load SAM Model
SAM_MODEL_PATH = "c:\\Users\\jyoth\\Downloads\\sam_vit_h_4b8939.pth"  # Update path
sam = sam_model_registry["vit_h"](checkpoint=SAM_MODEL_PATH)
predictor = SamPredictor(sam)

# Load OCR Model
reader = easyocr.Reader(["en"])

def process_ui_image(image_path):
    """Detect UI components using SAM and extract text using OCR."""
    image = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    predictor.set_image(image_rgb)
    masks, _, _ = predictor.predict(point_coords=None, point_labels=None, multimask_output=True)

    components = []
    for mask in masks:
        y, x = np.where(mask)
        if len(x) > 0 and len(y) > 0:
            x1, x2, y1, y2 = map(int, [min(x), max(x), min(y), max(y)])
            cropped = image[y1:y2, x1:x2]

            # OCR for text detection
            text_results = reader.readtext(cropped)
            text = text_results[0][-2] if text_results else "Unknown"

            components.append({
                "type": "button" if "button" in text.lower() else "text_field",
                "text": text,
                "position": {"x": x1, "y": y1},
                "size": {"width": x2 - x1, "height": y2 - y1}
            })

    return components

# Test the function
if __name__ == "__main__":
    image_path = "D:\\ui image 2.jpg"  # Replace with your UI image path
    ui_components = process_ui_image(image_path)

    # Save JSON output
    with open("ui_output.json", "w") as json_file:
        json.dump(ui_components, json_file, indent=2)

    print("✅ UI Components Saved to ui_output.json")
```

```python
import json
import requests

# Hugging Face API Key (Replace with your actual key)
HUGGINGFACE_API_KEY = "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"
HF_MODEL_NAME = "mistralai/Mistral-7B-Instruct-v0.1"  # Choose a suitable model

headers = {
    "Authorization": f"Bearer {HUGGINGFACE_API_KEY}",
    "Content-Type": "application/json"
}

def generate_ui_code(json_file):
    """Generate HTML, CSS, and JavaScript from UI JSON using Hugging Face API."""
    # Load JSON data
    try:
        with open(json_file, "r") as file:
            components = json.load(file)
    except Exception as e:
        print(f"❌ Error loading JSON file: {e}")
        return

    prompt = f"Convert these UI components into a responsive HTML, CSS, and JavaScript page:\n{json.dumps(components, indent=2)}"

    data = {
        "inputs": prompt,
        "parameters": {"max_new_tokens": 800, "temperature": 0.5}
    }

    # Send request to Hugging Face API
    response = requests.post(f"https://api-inference.huggingface.co/models/{HF_MODEL_NAME}", headers=headers, json=data)

    # Debug: Print response status
    print("🔹 API Response Status:", response.status_code)

    if response.status_code != 200:
        print(f"❌ Error: API Request Failed. Status Code: {response.status_code}")
        print("🔹 Response Text:", response.text)
        return

    # Parse API Response
    try:
        response_json = response.json()
        generated_code = response_json[0]["generated_text"]

        # Save output to an HTML file
        with open("ui_output.html", "w") as html_file:
            html_file.write(generated_code)
        
        print("✅ HTML Code Saved to ui_output.html")

    except json.JSONDecodeError:
        print("❌ Error: API response is not valid JSON.")
        print("🔹 Full Response:", response.text)


# Test the function
if __name__ == "__main__":
    generate_ui_code("ui_output.json")
```
